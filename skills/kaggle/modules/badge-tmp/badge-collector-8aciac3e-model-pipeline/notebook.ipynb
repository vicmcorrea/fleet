```python
import json

# Create a simple 'model' (just config/weights as JSON)
model = {
    'type': 'linear',
    'weights': [0.5, -0.3, 0.8],
    'bias': 0.1,
    'features': ['x1', 'x2', 'x3']
}

with open('model.json', 'w') as f:
    json.dump(model, f, indent=2)

print('Model saved to model.json')
print(json.dumps(model, indent=2))
```
