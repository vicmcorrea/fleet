```python
import pandas as pd

# Generate output data
data = pd.DataFrame({
    'x': range(1, 11),
    'y': [i**2 for i in range(1, 11)],
    'label': ['even' if i%2==0 else 'odd' for i in range(1, 11)]
})

# Save to output
data.to_csv('output_data.csv', index=False)
print(f'Output shape: {data.shape}')
print(data.head())
```
